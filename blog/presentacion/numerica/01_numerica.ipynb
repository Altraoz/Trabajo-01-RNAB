# Objetivo

Esta pagina resume la parte de optimizacion numerica del proyecto. La idea es mostrar el problema, el enfoque seguido y los resultados principales sin exponer todo el cuaderno de trabajo.


## Lo que se trabajo

- Definicion de funciones objetivo y criterios de comparacion.
- Analisis con metodos basados en gradiente.
- Comparacion final entre enfoques numericos.


## Resultados clave

Aqui puedes dejar las conclusiones mas importantes del componente numerico: que metodo convergio mejor, con que funcion se observaron mas dificultades y cuales fueron los patrones destacados en las corridas.
