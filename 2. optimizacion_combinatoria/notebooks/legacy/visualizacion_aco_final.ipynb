# Visualizacion del tour final ACO

Este notebook carga el mejor tour encontrado por ACO, muestra su convergencia y dibuja el recorrido final sobre un mapa de Francia.

```python
from pathlib import Path

from IPython.display import SVG, display

from visualizar_aco_final import (
    BASE_DIR,
    create_history_svg,
    create_route_svg,
    load_aco_result,
    load_city_names,
    load_expanded_tour,
    save_svg,
)

city_names = load_city_names()
aco_result = load_aco_result()
expanded_tour = load_expanded_tour()

best_tour = aco_result["mejor_tour_indices"]
history = aco_result["historial_mejor_costo"]

print(f"Costo final del tour: {aco_result['mejor_costo_tour']:.6f}")
print(f"Ciudades visitadas: {len(aco_result['mejor_tour_indices_sin_cierre'])}")
print(f"Saltos reales expandidos: {expanded_tour['n_saltos_reales']}")
print(f"Formula de costo: {aco_result['formula_peso']}")
print(f"Vehiculo de referencia: {aco_result['vehiculo_referencia']['name']}")
```

```python
history_svg = create_history_svg(history)
display(SVG(history_svg))
```

```python
route_svg = create_route_svg(
    best_tour,
    city_names,
    title="Tour final ACO sobre 96 capitales de Francia",
)
display(SVG(route_svg))
```

```python
output_route_svg = BASE_DIR / "fase_7_aco_final" / "visualizacion_aco_final.svg"
output_history_svg = BASE_DIR / "fase_7_aco_final" / "convergencia_aco_final.svg"
save_svg(route_svg, output_route_svg)
save_svg(history_svg, output_history_svg)
output_route_svg, output_history_svg
```
