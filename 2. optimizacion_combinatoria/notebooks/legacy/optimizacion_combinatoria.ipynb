```python
import pandas as pd
import numpy as np

# Leer CSV
df = pd.read_csv("dataset_francecities.csv")

# Asegurar que los índices sean enteros
df["indice_inicio"] = df["indice_inicio"].astype(int)
df["indice_destino"] = df["indice_destino"].astype(int)

# Limpiar columnas numéricas
columnas = ["distancia", "peajes(euros)", "gasolina(euros)"]

for col in columnas:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace("€", "", regex=False)
        .str.replace("km", "", regex=False)
        .str.replace(",", ".", regex=False)
        .str.strip()
    )

    df[col] = pd.to_numeric(df[col], errors="coerce")

# Tamaño de la matriz
n = max(df["indice_inicio"].max(), df["indice_destino"].max()) + 1

# Crear matriz cuadrada llena de infinito
matriz = np.full((n, n), np.inf)

# La diagonal debe ser 0 porque una ciudad hacia sí misma cuesta 0
np.fill_diagonal(matriz, 0)

# Llenar matriz con las conexiones existentes
for _, fila in df.iterrows():
    i = int(fila["indice_inicio"])
    j = int(fila["indice_destino"])

    valor = (
        fila["distancia"] +
        fila["peajes(euros)"] +
        fila["gasolina(euros)"]
    )

    # Si el valor es válido, se llena la matriz
    if not np.isnan(valor):
        matriz[i, j] = valor
        matriz[j, i] = valor  # simétrica

# Por seguridad, cualquier NaN que haya quedado se convierte en infinito
matriz[np.isnan(matriz)] = np.inf

print(matriz)
```

```python
import folium
import numpy as np

# Crear mapa
mapa = folium.Map(
    location=[Ciudades[:,0].mean(), Ciudades[:,1].mean()],
    zoom_start=5
)

# ====== PUNTOS ======
for i, punto in enumerate(Ciudades):

    folium.CircleMarker(
        location=[punto[0], punto[1]],
        radius=4,
        popup=f"Punto {i}",
        tooltip=f"{i}",
        fill=True
    ).add_to(mapa)

# ====== RUTA DEL OPTIMIZER SIN CERRAR CICLO ======

# Si best_path trae repetido el primer nodo al final, lo quitamos
best_path = list(optimizer.best_path)

if best_path[0] == best_path[-1]:
    best_path = best_path[:-1]

ruta = []

for idx in best_path:
    lat = Ciudades[idx][0]
    lon = Ciudades[idx][1]
    ruta.append([lat, lon])

# Dibujar línea siguiendo el orden, SIN unir final con inicio
folium.PolyLine(
    ruta,
    color="red",
    weight=4,
    opacity=0.8
).add_to(mapa)

# ====== MARCAR INICIO ======
inicio_idx = best_path[0]
inicio = Ciudades[inicio_idx]

folium.Marker(
    location=[inicio[0], inicio[1]],
    popup=f"INICIO - Punto {inicio_idx}",
    tooltip=f"INICIO: {inicio_idx}",
    icon=folium.Icon(color="green", icon="play")
).add_to(mapa)

# ====== MARCAR FINAL ======
final_idx = best_path[-1]
final = Ciudades[final_idx]

folium.Marker(
    location=[final[0], final[1]],
    popup=f"FINAL - Punto {final_idx}",
    tooltip=f"FINAL: {final_idx}",
    icon=folium.Icon(color="red", icon="stop")
).add_to(mapa)

# Mostrar mapa
mapa
```
