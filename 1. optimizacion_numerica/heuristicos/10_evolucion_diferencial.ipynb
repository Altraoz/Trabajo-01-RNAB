# Evolucion diferencial

Este notebook muestra una implementacion generica de evolucion diferencial para minimizacion y ejemplos rapidos de uso con funciones del proyecto.

```python
import sys
from pathlib import Path

ROOT_DIR = Path.cwd().parent
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

import matplotlib.pyplot as plt
import numpy as np

from funciones_heuristicas import run_differential_evolution
from funciones_objetivo import griewank, rastrigin, rosenbrock, schwefel
```

```python
CONFIG_ROSENBROCK_2D = {
    "objective_function": rosenbrock,
    "population_size": 40,
    "dimension": 2,
    "lower_bounds": -2.048,
    "upper_bounds": 2.048,
    "mutation_factor": 0.8,
    "crossover_rate": 0.7,
    "max_iterations": 100,
    "seed": 42,
}

CONFIG_SCHWEFEL_3D = {
    "objective_function": schwefel,
    "population_size": 50,
    "dimension": 3,
    "lower_bounds": -500.0,
    "upper_bounds": 500.0,
    "mutation_factor": 0.8,
    "crossover_rate": 0.7,
    "max_iterations": 120,
    "seed": 42,
}
```

```python
resultado_rosenbrock = run_differential_evolution(**CONFIG_ROSENBROCK_2D)
resultado_schwefel = run_differential_evolution(**CONFIG_SCHWEFEL_3D)

print("Rosenbrock 2D")
print("  Mejor solucion:", resultado_rosenbrock["best_solution"])
print("  Mejor valor:", resultado_rosenbrock["best_value"])
print()
print("Schwefel 3D")
print("  Mejor solucion:", resultado_schwefel["best_solution"])
print("  Mejor valor:", resultado_schwefel["best_value"])
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(resultado_rosenbrock["best_values_history"], color="seagreen")
axes[0].set_title("Mejor valor por iteracion - Rosenbrock 2D")
axes[0].set_xlabel("Iteracion")
axes[0].set_ylabel("Mejor valor")

axes[1].plot(resultado_schwefel["best_values_history"], color="darkorange")
axes[1].set_title("Mejor valor por iteracion - Schwefel 3D")
axes[1].set_xlabel("Iteracion")
axes[1].set_ylabel("Mejor valor")

plt.tight_layout()
plt.show()
```

```python
def probar_funcion(nombre: str, funcion, dimension: int, lower: float, upper: float, seed: int = 42):
    resultado = run_differential_evolution(
        objective_function=funcion,
        population_size=40,
        dimension=dimension,
        lower_bounds=lower,
        upper_bounds=upper,
        mutation_factor=0.8,
        crossover_rate=0.7,
        max_iterations=100,
        seed=seed,
    )
    print(nombre)
    print("  Mejor solucion:", resultado["best_solution"])
    print("  Mejor valor:", resultado["best_value"])
    return resultado
```
